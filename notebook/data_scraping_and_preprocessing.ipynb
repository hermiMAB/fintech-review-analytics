Web Scraping


```python

import sys
import os

#  Tell the notebook where to find your 'src' folder
sys.path.append(os.path.abspath('../'))
from src.preprocessing import get_app_metadata, scrape_bank_reviews, clean_reviews_dataframe, save_processed_data

# App IDs for the target Ethiopian banks [cite: 17, 40]
target_banks = {
    "Commercial Bank of Ethiopia": "com.combanketh.mobilebanking",
    "Bank of Abyssinia": "com.boa.boaMobileBanking",
    "Dashen Bank": "com.dashen.dashensuperapp"
}

# Fetch metadata for each bank [cite: 116, 121]
for bank_name, app_id in target_banks.items():
    get_app_metadata(app_id, bank_name)
```

```python
raw_dataframes = {}

for bank_name, app_id in target_banks.items():
    # We scrape slightly more than 400 to account for duplicates/empty rows [cite: 119, 44]
    df_raw = scrape_bank_reviews(app_id, bank_name, count=600)
    raw_dataframes[bank_name] = df_raw
```

Data Preprocessing and Quality Audit

```python
cleaned_dataframes = []

for bank_name, df_raw in raw_dataframes.items():
    print(f"Processing Audit for: {bank_name}")
    # clean_reviews_dataframe handles duplicates, missing values, and normalization 
    df_clean = clean_reviews_dataframe(df_raw)
    cleaned_dataframes.append(df_clean)
```

Dataset Finalization and Export

```python
import pandas as pd

# Save individual bank files 
for df in cleaned_dataframes:
    bank_label = df['bank'].iloc[0].lower().replace(' ', '_')
    save_processed_data(df, f"{bank_label}_reviews_clean.csv")

# Create and save the master combined dataset
master_df = pd.concat(cleaned_dataframes, ignore_index=True)
save_processed_data(master_df, "all_banks_combined_clean.csv")

print(f"\nCollected {len(master_df)} total clean reviews.")
```
